# LangChain Document Loading & Processing

This notebook demonstrates:
1. Creating `Document` objects manually
2. Loading a single text file with `TextLoader`
3. Loading all files in a directory using `DirectoryLoader` (loads both text files)
4. Loading PDF documents using `PyPDFLoader`

## 1. Creating Document Objects Manually

```python
from langchain_core.documents import Document

# Manually construct a Document
doc = Document(
    page_content="LangChain is a framework for developing applications powered by language models.",
    metadata={"source": "manual", "topic": "LangChain Intro"}
)

print("Document created successfully:")
print("Content:", doc.page_content)
print("Metadata:", doc.metadata)
```

## 2. Loading a Single Text File (`TextLoader`)
`TextLoader` targets **one specific file**.

```python
from langchain_community.document_loaders import TextLoader

# Load a single text file
loader = TextLoader("../data/text_files/python_intro.txt")
docs_single = loader.load()

print(f"Loaded {len(docs_single)} document:")
print(docs_single)
```

## 3. Loading Multiple Text Files (`DirectoryLoader`)
To load **both** `python_intro.txt` and `machine_learning.txt` from `../data/text_files/`, use `DirectoryLoader` with `glob="*.txt"` and `loader_cls=TextLoader`.

```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader

# Load all .txt files from the text_files folder
dir_loader = DirectoryLoader("../data/text_files", glob="*.txt", loader_cls=TextLoader)
all_text_docs = dir_loader.load()

print(f"Total documents loaded from directory: {len(all_text_docs)}")
for i, d in enumerate(all_text_docs, 1):
    print(f"\n--- Document {i} ---")
    print("Source:", d.metadata.get("source"))
    print("Preview:", d.page_content[:150], "...")
```

## 4. Loading PDF Documents (`PyPDFLoader`)
Extract page-by-page text and metadata from PDF files in `../data/pdf/`.

```python
from langchain_community.document_loaders import PyPDFLoader

# Load a PDF document
pdf_loader = PyPDFLoader("../data/pdf/01. project_overview.pdf")
pdf_pages = pdf_loader.load()

print(f"Total PDF pages loaded: {len(pdf_pages)}")
print("Page 1 Metadata:", pdf_pages[0].metadata)
print("\nPage 1 Content Preview:\n", pdf_pages[0].page_content[:200])
```
