# LangChain Document Basics

This notebook demonstrates how to create and use `Document` objects from `langchain_core`.

```python
from langchain_core.documents import Document

# Create a Document object
doc = Document(
    page_content="LangChain is a framework for developing applications powered by language models.",
    metadata={"source": "intro.txt", "author": "LangChain", "page": 1}
)

print("Document created successfully:")
print(doc)
```

```python
# Access content and metadata
print("Page Content:", doc.page_content)
print("Metadata:", doc.metadata)
```

### Creating a List of Documents

```python
docs = [
    Document(
        page_content="Retrieval-Augmented Generation (RAG) combines search with generative LLMs.",
        metadata={"topic": "RAG", "id": 1}
    ),
    Document(
        page_content="Vector stores index embeddings for efficient semantic similarity lookup.",
        metadata={"topic": "Vector Stores", "id": 2}
    )
]

for i, d in enumerate(docs, 1):
    print(f"Document {i} [{d.metadata['topic']}]: {d.page_content}")
```
