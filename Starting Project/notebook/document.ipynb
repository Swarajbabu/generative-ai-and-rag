```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader

dir_loader = DirectoryLoader(
    "../data/text_files",
    glob = "**/*.txt",
    loader_cls = TextLoader,
    loader_kwargs = {'encoding': 'utf-8'},
    show_progress = True
)
document1 = dir_loader.load()
print(document1)
```

```python
from langchain_community.document_loaders import PyMuPDFLoader, PyPDFLoader, DirectoryLoader

dir_loder = DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader,
    show_progress=True
)

pdf_loader = dir_loder.load()
print(pdf_loader)
```

```python
from langchain_community.document_loaders import UnstructuredHTMLLoader, UnstructuredURLLoader

loader = UnstructuredURLLoader(urls=['https://github.com/Swarajbabu/generative-ai-and-rag/tree/main'], mode='elements')
data_html = loader.load()

print(len(data_html))

print(data_html[0].page_content)
print(data_html[1].page_content)
print(data_html[2].page_content)

print(data_html[0].metadata)
```

```python
# Character-based chunking   

from langchain_text_splitters import CharacterTextSplitter

splitter = CharacterTextSplitter(
    separator="\n\n",       # Paragraph
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(pdf_loader)
# print(chunks)
print(len(chunks))
print(chunks[0])
```

```python
# 3. Recursive character text splitting

from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)
chunks = splitter.split_documents(pdf_loader)
print(chunks)
print(len(chunks))
print(chunks[0].page_content)
print(chunks[0].metadata)
```

# EmbeddingManager

```python
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
sentences = [
    "I love programming.",
    "I enjoy coding.",
    "Python is used for AI.",
    "The weather is very hot."
]

embeddings = model.encode(sentences)

print("Shape:", embeddings.shape)
```

```python
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
# sentence = "Artificial Intelligence is transforming the world."
chunk_texts = []
for chunk in chunks:
    chunk_texts.append(chunk.page_content)

embedding = model.encode(chunk_texts)

print(embedding)
```

```python
print("Embedding dimensions:", embedding.shape)
print("Total numbers:", len(embedding))
```
