```python
import chromadb

client_db = chromadb.PersistentClient(
    path="./chroma_db"
)
```

```python
collection = client_db.get_collection(
    name="My_rag_collections",
    embedding_function=None
)

print("Total stored chunks:", collection.count())
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "BAAI/bge-small-en-v1.5"
)
```

```python
from openai import OpenAI
import os

groq_client = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"]
)

MODEL_NAME = "openai/gpt-oss-20b"

print("Groq LLM connected!")
```

```python
def retrieve_documents(question):

    query_instruction = (
        "Represent this sentence for searching relevant passages: "
    )

    # Convert question into vector
    query_embedding = embedding_model.encode(
        query_instruction + question,
        normalize_embeddings=True
    )

    # Search ChromaDB
    results = collection.query(

        query_embeddings=[
            query_embedding.tolist()
        ],

        n_results=min(4, collection.count()),

        include=[
            "documents",
            "metadatas",
            "distances"
        ]
    )

    return results
```

```python
results = retrieve_documents(
    "Explain the main features of Wanderlust."
)

print(results["documents"][0])
```

## Prepare the retrieved context

```python
def prepare_context(results):

    context = ""

    documents = results["documents"][0]
    metadatas = results["metadatas"][0]

    for i, document in enumerate(documents):

        source = metadatas[i].get(
            "source", "Unknown"
        )

        page = metadatas[i].get("page", None)

        if page is not None:
            page = page + 1

        context += f"""
        Source {i+1}:
        File: {source}
        Page: {page}

        Content:
        {document}

        ----------------------------
        """

    return context
```

```python
def generate_answer(question, context):

    prompt = f"""
    Answer the user's question using the
    following document information.

    DOCUMENT CONTEXT:
    {context}

    USER QUESTION:
    {question}

    INSTRUCTIONS:

    1. Answer in simple and clear English.
    2. Use only the provided document context.
    3. Explain important points using bullet points.
    4. Use headings when needed.
    5. If the answer is not available in the
       documents, say "I could not find this
       information in the provided documents."
    6. Do not invent information.
    """

    response = groq_client.chat.completions.create(

        model=MODEL_NAME,

        messages=[
            {
                "role": "system",
                "content": "You are a helpful RAG chatbot."
            },
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    answer = response.choices[0].message.content

    return answer
```

```python
from IPython.display import display, Markdown

def display_answer(answer):

    display(
        Markdown(
            f"""
            # 🤖 AI Assistant

            {answer}
            """
        )
    )
```

```python
def ask_question(question):

    # 1. Retrieve relevant documents
    results = retrieve_documents(question)

    # 2. Prepare context
    context = prepare_context(results)

    # 3. Generate answer using LLM
    answer = generate_answer(
        question,
        context
    )

    # 4. Display answer
    display_answer(answer)

    return answer
```

## Asking the questions

```python
while True:

    question = input("\nYou: ")

    if question.lower() in ["exit", "quit", "bye"]:

        print("Chatbot: Goodbye!")
        break

    try:

        print("\nAI Assistant:")

        answer = ask_question(question)

    except Exception as e:

        print("Something went wrong:", e)
```
