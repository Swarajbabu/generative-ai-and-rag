```python
import pandas as pd

data = {
    "Age": [18, 19, 20, 21, 22, 23, 24, 25],
    "Salary": [20000, 22000, 25000, 30000,
               35000, 40000, 45000, 50000],
    "Experience": [0, 1, 1, 2, 3, 4, 5, 6],
    "Department": ["IT", "HR", "IT", "Sales",
                   "HR", "IT", "Sales", "IT"]
}

df = pd.DataFrame(data)

print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.barplot(
    data=df,
    x="Department",
    y="Salary"
)

plt.title("Average Salary by Department")
plt.show()
```

```python
sns.countplot(
    data=df,
    x="Department"
)

plt.title("Department Count")
plt.show()
```

```python
sns.boxplot(
    data=df,
    y="Salary"
)

plt.title("Salary Distribution")
plt.show()
```

```python
correlation = df[
    ["Age", "Salary", "Experience"]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Feature Correlation")
plt.show()
```

```python
sns.pairplot(
    df,
    hue="Department"
)

plt.show()
```
